# **Mine candidate terms from the data**

```python
STOP = set("a an and or of the to for with in on at into from by as is are be it its if than then plus more about "
           "each any all some your you will can may not no".split())
UNITS = set("cup cups tablespoon tablespoons tbsp tbsps tsp tsps teaspoon teaspoons oz ounce ounces lb lbs pound pounds "
            "g gram grams kg ml liter liters can cans package packages pkg pinch dash large small medium whole fresh "
            "chopped sliced diced minced ground cut finely thinly cooked divided optional taste inch pieces piece "
            "clove cloves lightly freshly packed softened melted drained rinsed peeled".split())

def parse_list(x):
    """ingredients may be a python-list string, a list, or a delimited string: handle all."""
    if isinstance(x, (list, tuple)): return list(x)
    try:
        v = ast.literal_eval(x)
        return list(v) if isinstance(v, (list, tuple)) else [str(v)]
    except Exception:
        return re.split(r"[|;\n]", str(x))

def clean_tokens(s):
    s = re.sub(r"[^a-z\s'-]", " ", str(s).lower())
    return [t for t in s.split() if t not in STOP and t not in UNITS and len(t) > 2]

def term_candidates(series, n=50000, top=800, bigrams=True, is_list=True):
    sub = series.dropna().sample(min(n, series.notna().sum()), random_state=SEED)
    c = collections.Counter()
    for x in sub:
        items = parse_list(x) if is_list else [x]
        seen = set()
        for item in items:
            toks = clean_tokens(item)
            seen.update(toks)
            if bigrams: seen.update(a + " " + b for a, b in zip(toks, toks[1:]))
        c.update(seen)
    return pd.DataFrame(c.most_common(top), columns=["term", "n_docs"])

def phrase_after(texts, pattern, top=60):
    """e.g. r'\bfor my ([a-z]+)' finds what people cook 'for my ...'"""
    rx = re.compile(pattern, re.I); c = collections.Counter()
    for t in texts:
        for m in rx.finditer(t): c[m.group(1).lower()] += 1
    return pd.DataFrame(c.most_common(top), columns=["phrase", "count"])

ing_cand   = term_candidates(df_recipes["ingredients"], is_list=True)
dish_cand  = term_candidates(df_recipes["title"], is_list=False, top=500)
rev_sample = reviews["text"].sample(min(100000, len(reviews)), random_state=SEED)
diner_cand = phrase_after(rev_sample, r"\bfor (?:my|our) ([a-z]+)")
occ_cand   = phrase_after(rev_sample, r"\bfor (?:a|an|the|our|my) ((?:[a-z]+ )?(?:party|dinner|lunch|brunch|night|day|weekend|holiday|occasion|picnic|potluck|gathering))")

for name, d in [("ingredient", ing_cand), ("dish", dish_cand), ("diner_group", diner_cand), ("occasion", occ_cand)]:
    d.to_csv(f"candidates_{name}.csv", index=False)
    print(f"--- {name}: top 15 ---"); print(d.head(15).to_string(index=False))

# Baseline v0: single-word ingredient candidates, auto-accepted (noisy on purpose; curate to improve)
LEXICONS["INGREDIENT"] = sorted(set(LEXICONS["INGREDIENT"]) | set(t for t in ing_cand["term"].head(400) if " " not in t))
print("INGREDIENT lexicon size:", len(LEXICONS["INGREDIENT"]))
```

# **Dictionary baseline**

```python
PRIORITY = ["DINER_GROUP", "OCCASION", "COOKING_EQUIPMENT", "DURATION", "INGREDIENT", "DISH"]

class DictTagger:
    def __init__(self, lexicons):
        self.nlp = spacy.blank("en")
        self.matcher = PhraseMatcher(self.nlp.vocab, attr="LOWER")
        for label, terms in lexicons.items():
            if label == "DURATION": continue
            self.matcher.add(label, [self.nlp.make_doc(t) for t in sorted(set(terms))])

    def __call__(self, text):
        doc = self.nlp.make_doc(text)
        spans = []
        for mid, s, e in self.matcher(doc):
            sp = doc[s:e]
            spans.append((sp.start_char, sp.end_char, self.nlp.vocab.strings[mid]))
        spans += [(m.start(), m.end(), "DURATION") for m in DUR_RE.finditer(text)]
        # longest span first, then label priority, then position; drop overlaps
        spans.sort(key=lambda x: (-(x[1] - x[0]), PRIORITY.index(x[2]), x[0]))
        kept = []
        for s in spans:
            if all(s[1] <= k[0] or s[0] >= k[1] for k in kept): kept.append(s)
        return sorted(kept)

tagger = DictTagger(LEXICONS)

demo = "Made this in my slow cooker for my kids on a weeknight, only 20 minutes of prep, and the chicken was great."
print([(demo[s:e], l) for s, e, l in tagger(demo)])
```

# **Loading annotations and evaluate**

```python
def load_label_studio(path):
    gold = []
    for t in json.load(open(path)):
        anns = [a for a in t.get("annotations", []) if not a.get("was_cancelled")]
        if not anns: continue
        ents = [(r["value"]["start"], r["value"]["end"], r["value"]["labels"][0])
                for r in anns[0]["result"] if r.get("type") == "labels"]
        gold.append({"id": t["data"].get("review_id"), "text": t["data"]["text"], "entities": sorted(set(ents))})
    return gold

def _overlap(a, b): return a[0] < b[1] and b[0] < a[1]

def evaluate(gold, predict_fn, relaxed=False):
    """Entity-level P/R/F1. Strict = exact span + label. Relaxed = any overlap + same label."""
    stats = {l: {"tp": 0, "fp": 0, "fn": 0} for l in LABELS}
    for ex in gold:
        g = set(map(tuple, ex["entities"])); p = set(predict_fn(ex["text"]))
        for l in LABELS:
            gl = [x for x in g if x[2] == l]; pl = [x for x in p if x[2] == l]
            if relaxed:
                tp_p = sum(any(_overlap(x, y) for y in gl) for x in pl)
                tp_g = sum(any(_overlap(x, y) for y in pl) for x in gl)
                stats[l]["tp"] += tp_g; stats[l]["fp"] += len(pl) - tp_p; stats[l]["fn"] += len(gl) - tp_g
            else:
                tp = len(set(gl) & set(pl))
                stats[l]["tp"] += tp; stats[l]["fp"] += len(pl) - tp; stats[l]["fn"] += len(gl) - tp
    rows = []
    def prf(tp, fp, fn):
        p = tp / (tp + fp) if tp + fp else 0.0; r = tp / (tp + fn) if tp + fn else 0.0
        return p, r, (2 * p * r / (p + r) if p + r else 0.0)
    for l in LABELS:
        s = stats[l]; p, r, f = prf(s["tp"], s["fp"], s["fn"])
        rows.append({"label": l, "P": p, "R": r, "F1": f, "support": s["tp"] + s["fn"]})
    T = {k: sum(stats[l][k] for l in LABELS) for k in ("tp", "fp", "fn")}
    p, r, f = prf(**T); rows.append({"label": "MICRO", "P": p, "R": r, "F1": f, "support": T["tp"] + T["fn"]})
    mac = sum(x["F1"] for x in rows[:-1]) / len(LABELS)
    rows.append({"label": "MACRO-F1", "P": None, "R": None, "F1": mac, "support": None})
    return pd.DataFrame(rows).round(3)

# After you have annotated and exported (ls_export.json):
# gold = load_label_studio("ls_export.json")
# print(evaluate(gold, tagger))                 # strict
# print(evaluate(gold, tagger, relaxed=True))   # relaxed (boundary errors forgiven)
```

# **Training spaCy NER model and compare**

```python
def split_gold(gold, dev_frac=0.2, seed=SEED):
    g = gold[:]; random.Random(seed).shuffle(g)
    k = int(len(g) * (1 - dev_frac)); return g[:k], g[k:]

def make_examples(nlp_, data):
    exs = []
    for ex in data:
        doc = nlp_.make_doc(ex["text"]); ents = []
        for s, e, l in ex["entities"]:
            sp = doc.char_span(s, e, label=l, alignment_mode="expand")   # snaps to token boundaries
            if sp is not None: ents.append((sp.start_char, sp.end_char, l))
        exs.append(Example.from_dict(doc, {"entities": ents}))
    return exs

def train_spacy_ner(train, n_iter=30, batch=8, drop=0.3, base=None, seed=SEED):
    random.seed(seed)
    nlp_ = spacy.load(base, exclude=["ner"]) if base else spacy.blank("en")   # base="en_core_web_sm" for stronger features
    ner = nlp_.add_pipe("ner")
    for l in LABELS: ner.add_label(l)
    exs = make_examples(nlp_, train)
    nlp_.initialize(lambda: exs)
    for i in range(n_iter):
        random.shuffle(exs); losses = {}
        for b in spacy.util.minibatch(exs, size=batch): nlp_.update(b, drop=drop, losses=losses)
        if (i + 1) % 10 == 0: print(f"iter {i+1}: loss {losses['ner']:.1f}")
    return nlp_

def spacy_predictor(nlp_):
    return lambda text: [(e.start_char, e.end_char, e.label_) for e in nlp_(text).ents]

def show_errors(gold, predict_fn, label, k=8):
    fp, fn = [], []
    for ex in gold:
        g = {x for x in map(tuple, ex["entities"]) if x[2] == label}
        p = {x for x in predict_fn(ex["text"]) if x[2] == label}
        fp += [(ex["text"][s:e], ex["text"][max(0, s-30):e+30]) for s, e, _ in p - g]
        fn += [(ex["text"][s:e], ex["text"][max(0, s-30):e+30]) for s, e, _ in g - p]
    print(f"[{label}] false positives:"); [print("  ", a, "|", b.replace("\n", " ")) for a, b in fp[:k]]
    print(f"[{label}] missed (false negatives):"); [print("  ", a, "|", b.replace("\n", " ")) for a, b in fn[:k]]

# Once you have >= ~300 annotated reviews:
# train, dev = split_gold(gold)
# model = train_spacy_ner(train)                       # or base="en_core_web_sm"
# comp = pd.concat({"dictionary": evaluate(dev, tagger).set_index("label")["F1"],
#                   "spaCy NER":  evaluate(dev, spacy_predictor(model)).set_index("label")["F1"]}, axis=1)
# print(comp)
# show_errors(dev, tagger, "INGREDIENT"); show_errors(dev, spacy_predictor(model), "OCCASION")
```
